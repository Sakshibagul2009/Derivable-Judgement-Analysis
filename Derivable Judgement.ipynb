## Part A – Theoretical Foundation

### 1. What is Inferential Statistics?

Inferential statistics is a branch of statistics used to make conclusions
or predictions about a population based on sample data.

It includes methods such as hypothesis testing, confidence intervals,
and estimation.

### 2. What is Hypothesis Testing and its Components?

Hypothesis testing is a statistical method used to determine whether
there is enough evidence to support a claim about a population.

Its main components are Null Hypothesis (H₀), Alternative Hypothesis (H₁),
significance level, test statistic, p-value, critical value, and final decision.

### 3. Explain Confidence Interval and Critical Value

A confidence interval is a range of values within which a population
parameter is expected to lie with a specified confidence level.

A critical value is a boundary value used to determine whether the
test statistic falls in the rejection region.

### 4. Define p-value

A p-value represents the probability of obtaining a result as extreme
as the observed result, assuming that the null hypothesis is true.

If p-value is less than 0.05, the null hypothesis is generally rejected.

### 5. Differentiate Type I and Type II Errors

Type I Error occurs when a true null hypothesis is rejected.

Type II Error occurs when a false null hypothesis is not rejected.

### 6. Brief Description of Z-test, t-test, Chi-square Test and ANOVA

**Z-test:** Used mainly for testing means or proportions when the sample
size is large and population variance is known.

**t-test:** Used to compare the means of two groups, especially when
population standard deviation is unknown.

**Chi-square test:** Used to determine whether two categorical variables
are associated.

**ANOVA:** Used to determine whether there is a significant difference
among the means of three or more groups.

### 7. What is Covariance?

Covariance measures the direction in which two numerical variables
change together.

A positive covariance indicates that the variables tend to increase
together, while a negative covariance indicates an opposite movement.

### 8. What is Correlation?

Correlation measures the strength and direction of the relationship
between two numerical variables.

The Pearson correlation coefficient ranges from -1 to +1.
A value close to +1 indicates a strong positive relationship,
while a value close to -1 indicates a strong negative relationship.

## 1. Formulation of Hypotheses

Based on the health dataset, the following hypotheses are formulated.

### Hypothesis 1: Smoking Status and Diabetes

**H₀:** Smoking status has no significant association with diabetes.

**H₁:** Smoking status has a significant association with diabetes.

### Hypothesis 2: Glucose Level and Diabetes

#  2: Calculate 95% Confidence Interval

A 95% confidence interval is calculated for age, weight, BMI,
glucose level, and blood pressure.

The formula is:

CI = Mean ± Critical Value × Standard Error

```python
# 95% Confidence Interval

alpha = 0.05

variables = [
    "age",
    "weight",
    "bmi",
    "glucose_level",
    "blood_pressure"
]

ci_results = []

for variable in variables:

    data = df[variable].dropna()

    n = len(data)
    mean = data.mean()
    std = data.std(ddof=1)

    standard_error = std / np.sqrt(n)

    critical_value = stats.t.ppf(
        1 - alpha / 2,
        n - 1
    )

    lower = mean - critical_value * standard_error
    upper = mean + critical_value * standard_error

    ci_results.append([
        variable,
        n,
        mean,
        std,
        standard_error,
        critical_value,
        lower,
        upper
    ])

ci_df = pd.DataFrame(
    ci_results,
    columns=[
        "Variable",
        "N",
        "Mean",
        "Standard Deviation",
        "Standard Error",
        "Critical Value",
        "CI Lower",
        "CI Upper"
    ]
)

ci_df
```

The 95% confidence intervals for the selected numerical variables were
successfully calculated.

The results are:

- **Age:** 95% CI = 49.2395 to 51.5205
- **Weight:** 95% CI = 67.0654 to 68.5986
- **BMI:** 95% CI = 24.2489 to 24.9229
- **Glucose Level:** 95% CI = 104.9363 to 106.9641
- **Blood Pressure:** 95% CI = 146.3615 to 147.9335

The critical t-value at the 95% confidence level is approximately
**1.9623**.

Therefore, these intervals represent the range of plausible population
means at the 95% confidence level.

## 3. Find the Critical Value and p-value

Find the critical value and p-value for the statistical tests and use them
to determine whether the null hypothesis should be rejected or not.

```python
alpha = 0.05

# 1. Critical value for t-test
critical_t = stats.t.ppf(1 - alpha/2, df=len(df) - 2)

print("Critical t-value:", critical_t)

# 2. Critical value for Chi-square test
critical_chi2 = stats.chi2.ppf(1 - alpha, df=2)

print("Critical Chi-square value:", critical_chi2)

# 3. Critical value for ANOVA
critical_f = stats.f.ppf(1 - alpha, dfn=4, dfd=len(df) - 5)

print("Critical F-value:", critical_f)
```

### Interpretation

At the 5% significance level:

- For the **t-test**, the p-value is less than 0.05. Therefore, we reject H₀.
- For the **Chi-square test**, the p-value is greater than 0.05. Therefore, we fail to reject H₀.
- For the **ANOVA test**, the p-value is less than 0.05. Therefore, we reject H₀.

Thus, the critical values and p-values help determine whether the
observed results are statistically significant.

## 4. Perform t-test

Compare the mean glucose levels between diabetic and non-diabetic
individuals using an independent two-sample t-test.

- **H₀:** There is no significant difference in mean glucose levels between diabetic and non-diabetic individuals.
- **H₁:** There is a significant difference in mean glucose levels between diabetic and non-diabetic individuals.

Formula:

t = (x̄₁ - x̄₂) / √(s₁²/n₁ + s₂²/n₂)

```python
diabetic = df[df["diabetes"] == True]["glucose_level"].dropna()
non_diabetic = df[df["diabetes"] == False]["glucose_level"].dropna()

t_stat, p_value = stats.ttest_ind(
    diabetic,
    non_diabetic,
    equal_var=False
)

print("Diabetic Mean:", diabetic.mean())
print("Non-Diabetic Mean:", non_diabetic.mean())
print("t-statistic:", t_stat)
print("p-value:", p_value)
```

### Interpretation

The independent two-sample t-test was performed to compare the mean
glucose levels between diabetic and non-diabetic individuals.

- **Diabetic Mean:** 131.4482
- **Non-Diabetic Mean:** 99.7737
- **t-statistic:** 37.8708
- **p-value:** 3.2644 × 10⁻¹¹⁵

Since the p-value is much less than 0.05, we reject the null hypothesis (H₀).

Therefore, there is a statistically significant difference in mean
glucose levels between diabetic and non-diabetic individuals.

## 5. Conduct a Chi-square Test

Conduct a Chi-square test to determine whether there is a significant
association between smoking status and diabetes.

- **H₀:** Smoking status has no significant association with diabetes.
- **H₁:** Smoking status has a significant association with diabetes.

```python
# Create contingency table
contingency_table = pd.crosstab(
    df["smoking_status"],
    df["diabetes"]
)

print("Contingency Table:")
print(contingency_table)

# Perform Chi-square test
chi2_stat, p_value, degrees_of_freedom, expected = stats.chi2_contingency(
    contingency_table
)

# Critical value at 5% significance level
alpha = 0.05
critical_value = stats.chi2.ppf(
    1 - alpha,
    degrees_of_freedom
)

print("\nChi-square statistic:", chi2_stat)
print("Degrees of freedom:", degrees_of_freedom)
print("Critical value:", critical_value)
print("p-value:", p_value)
```

### Interpretation

The Chi-square test was conducted to determine whether there is a
significant association between smoking status and diabetes.

- **Chi-square statistic:** 2.091
- **Degrees of freedom:** 2
- **Critical value:** 5.991
- **p-value:** 0.3515

Since the p-value is greater than 0.05, we fail to reject the null
hypothesis (H₀).

Therefore, there is no statistically significant association between
smoking status and diabetes in this dataset.

### 6. Perform an ANOVA Test

Perform a one-way ANOVA test to determine whether diabetes rates
differ significantly across different age groups.

- **H₀:** There is no significant difference in diabetes rates across age groups.
- **H₁:** There is a significant difference in diabetes rates across age groups.

Formula:

F = MS Between / MS Within

```python
# Convert diabetes to numerical values
df["diabetes_numeric"] = df["diabetes"].astype(int)

# Create groups based on age_group
groups = [
    group["diabetes_numeric"].dropna()
    for name, group in df.groupby("age_group")
]

# Perform one-way ANOVA
f_stat, p_value = stats.f_oneway(*groups)

# Degrees of freedom
k = len(groups)
N = len(df)

df_between = k - 1
df_within = N - k

# Critical F-value
alpha = 0.05
critical_f = stats.f.ppf(
    1 - alpha,
    df_between,
    df_within
)

print("F-statistic:", f_stat)
print("p-value:", p_value)
print("Degrees of freedom (Between):", df_between)
print("Degrees of freedom (Within):", df_within)
print("Critical F-value:", critical_f)
```

### Interpretation

The one-way ANOVA test was performed to determine whether diabetes
rates differ significantly across different age groups.

- **F-statistic:** 25.821
- **p-value:** 2.19 × 10⁻²⁰
- **Degrees of freedom (Between):** 4
- **Degrees of freedom (Within):** 995
- **Critical F-value:** 2.381

Since the p-value is much less than 0.05, we reject the null hypothesis
(H₀).

Therefore, there is a statistically significant difference in diabetes
rates across the different age groups.

## 7. Calculate Covariance and Correlation

Calculate the covariance and Pearson correlation between age and BMI.

- **H₀:** There is no significant correlation between age and BMI.
- **H₁:** There is a significant correlation between age and BMI.

```python
# Select Age and BMI
age_bmi = df[["age", "bmi"]].dropna()

# Covariance
covariance = age_bmi["age"].cov(age_bmi["bmi"])

# Pearson correlation
correlation, p_value = stats.pearsonr(
    age_bmi["age"],
    age_bmi["bmi"]
)

print("Covariance:", covariance)
print("Pearson Correlation:", correlation)
print("p-value:", p_value)
```

### Interpretation

The covariance and Pearson correlation between age and BMI were
calculated.

- **Covariance:** 4.231
- **Pearson Correlation (r):** 0.0424
- **p-value:** 0.1804

The correlation coefficient is close to 0, indicating a very weak
positive relationship between age and BMI.

Since the p-value is greater than 0.05, we fail to reject the null
hypothesis (H₀).

Therefore, there is no statistically significant correlation between
age and BMI in this dataset.
